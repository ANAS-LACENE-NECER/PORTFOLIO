#### Question 1 : Nombre d'amis d'une personne

- Étant donné un tableau `amis`, écrire une fonction `nb_amis(amis, prenom)` qui retourne le nombre d'amis de `prenom` à partir des données du tableau `amis`. 

````python
def nb_amis(amis,prenom):                        #Définition de la fonction
    """
    Compte le nombre de fois qu'un prénom apparaît dans la liste `amis`, et ainsi compter le nombre d'amis que 
    ce prénom a.

    Arguments :
    - amis : Tableau des relation d'amitié
    - prenom : le prénom dont on souhaite compter le nombre d'amis.

    Retourne :
    - le nombre d'amis de prenom à partir des données du tableau amis
    """    
    
    i=0                                           #Initialisation de i qui permet de parcourire la liste
    compte=0                                      #Initialisation du compteur d'amitiés
    while i<len(amis):                            #Parcours de toute la liste `amis`
        if amis[i]==prenom:                       #Si le prénom dont on souhaite connaître son nombre d'amis est égal au prénom cherché
            compte+=1                             #Incrémentation du compteur
        i+=1                                      #Passage au prénom suivant
    return compte

p_amis = ['Muriel','Yasmine','Muriel','Joël',
          'Yasmine','Joël','Yasmine','Thomas',
          'Daria','Thomas','Carole','Daria',
          'Joël','Nassim','Joël','Andrea',
          'Joël','Ali','Nassim','Andrea',
          'Nassim','Ali','Andrea','Ali',
          'Valentin','Andrea','Leo',
          'Valentin','Thierry','Axel',
          'Thierry','Leo','Axel','Leo']           #Tableau utilisé pour le test de la fonction 

print(nb_amis(p_amis,'Yasmine'))                  #Affichage du resultat ('résultat attendu : 3') pour Yasmine 




````

#### Question 2 : Nombre de membres d'un réseau social 

- Ecrire une fonction `taille_reseau(amis)` qui à partir d'un tableau `amis`retourne le nombre de personnes distinctes participant à ce réseau social.

````python
def taille_reseau(amis):

    """
    Calcule la taille du réseau d'amitié en comptant le nombre unique de personnes dans la liste `amis`.

    Arguments :
    - amis : liste contenant des prénoms en paires, chaque prénom représentant une personne impliquée dans une amitié.

    Retourne :
    - le nombre de personnes distinctes participant à ce réseau social.
    """
    i=0
    t=[]                                          #Inition d'un tableau pour stocker les prénoms uniques rencontrés
    while i<len(amis):
        if amis[i] not in t:                      #Si le prénom courant n'est pas déjà dans t
            t.append(amis[i])                     #Ajout du prénom dans le tableau des prénoms uniques
        i+=1
    return len(t)
p_amis = ['Muriel','Yasmine','Muriel','Joël',
          'Yasmine','Joël','Yasmine','Thomas',
          'Daria','Thomas','Carole','Daria',
          'Joël','Nassim','Joël','Andrea',
          'Joël','Ali','Nassim','Andrea',
          'Nassim','Ali','Andrea','Ali',
          'Valentin','Andrea','Leo',
          'Valentin','Thierry','Axel',
          'Thierry','Leo','Axel','Leo'] 

print(taille_reseau(p_amis))                      #Résultat attendu : 13




````

#### Question 3 :  Lecture des données d'un réseau à partir d'un fichier
On suppose que les données sur un réseau social sont stockées dans un fichier CSV de la manière suivante :
```
prenom1;prenom2
prenom3;prenom4
prenom5;prenom6
...
```
Autrement dit, chaque ligne du fichier contient une paire de prénoms séparés par un `';'` correspondant à deux personnes ayant des interactions. 

**NB** : Quatre fichiers CSV de ce type sont fournis dans le répertoire `files/`, il s'agit des fichiers`Communaute1.csv`, `Communaute2.csv`,`Communaute3.csv` et `Communaute4.csv`. 
Il est recommandé d'en fabriquer d'autres. 

- Ecrire une fonction `lecture_reseau(path)` prenant en paramètre un chemin vers un tel fichier CSV et retournant un tableau modélisant les interactions entre les personnes du fichier.

````python
def lecture_reseau(path):
    """
    Lit un fichier CSV contenant des relations d'amitié et retourne une liste de listes,
    où chaque sous-liste contient les noms des personnes en relation d'amitié.

    Arguments :
    - path : chemin d'accès au fichier CSV (chaîne de caractères).
      Le fichier est supposé contenir des lignes au format 'nom1;nom2', représentant une amitié entre 'nom1' et 'nom2'.

    Retourne :
    - un tableau modélisant les interactions entre les personnes du fichier.

    """

    
    fichier = open(path, "r", encoding="utf-8")        # Ouverture du fichier en mode lecture avec encodage UTF-8
    a = fichier.readlines()                            # Lecture de toutes les lignes du fichier
    i = 0
    tab = []                                           # Initialisation d'un tableau pour stocker les relations d'amitié

  
    while i < len(a):
        ligne = a[i]                                   # Récupération de la ligne i
        ln = ligne.strip().split(';')                  # Suppression des '/n' et remplacement des ';' en ','et fait un tableau
        tab.append(ln)                                 # Ajout de la relation d'amitié dans le tableau
        i += 1  

    fichier.close()                                    # Fermeture du fichier après lecture
    return tab

path = "files/Communaute1.csv"                         # Chemin d'accès au fichier CSV contenant le réseau d'amitié

print(lecture_reseau(path))                            # Affichage de la liste des relations d'amitié





````

#### Question  4 :  Données d'un réseau dupliquées dans un fichier
On suppose toujours que les données sur un réseau social sont stockées dans un fichier CSV mais certaines lignes du fichier contiennent les mêmes informations.

- Ecrire une fonction `lecture_reseau_bis(path)` prenant en paramètre un chemin vers un tel fichier CSV et retournant un tableau modélisant les interactions entre les personnes du fichier en éliminant les répétitions.

````python
def lecture_reseau_bis(path):
    """
    Lit un fichier CSV contenant des relations d'amitié et retourne un tableau de tableau uniques,
    où chaque sous-tableau contient deux noms représentant une relation d'amitié unique.

    Arguments :
    - path : chemin d'accès au fichier CSV (chaîne de caractères).

    Retourne :
    - un tableau modélisant les interactions entre les personnes du fichier en éliminant les répétitions.

    """

    
    fichier = open(path, "r", encoding="utf-8")
    a = fichier.readlines() 
    i = 0
    tab = [] 

    while i < len(a):
        ligne = a[i]  
        ln = ligne.strip().split(';')
        
        if ln not in tab:  # Vérification pour éviter les doublons
            tab.append(ln)  
        
        i += 1 

    fichier.close() 
    return tab


path = "filesbis/Communaute1bis.csv"
print(lecture_reseau_bis(path))  # Affichage des relations d'amitié uniques





````

#### Question 5 : Modélisation d'un réseau par un dictionnaire
On préfère pour la suite, utiliser une modélisation du réseau social par un dictionnaire dont les clés sont les prénoms des personnes du réseau et la valeur associé à une clé est le tableau des amis de la personne indiquée par la clé.

- A partir d'un tableau `amis` modélisant les interactions entre personnes d'un réseau, écrire une fonction `dico_reseau(amis)` qui retourne un dictionnaire dont les clés sont les prénoms des membres du réseau et les valeurs le tableau de leurs amis.

````python
def dico_reseau(amis):
    """
    Construit un dictionnaire représentant un réseau d'amitié à partir d'une liste de paires d'amis.

    Arguments :
    - amis : liste de paires d'amis (liste de listes ou de tuples de deux noms).
      Chaque paire représente une relation bidirectionnelle entre deux amis.

    Retourne :
    - un dictionnaire dont les clés sont les prénoms des membres du réseau et les valeurs le tableau de leurs amis.
   
    """

    dico_reseau = {}  # Initialisation du dictionnaire pour stocker le réseau d'amitié

    # Parcourir chaque duo d'amis dans la liste
    for duo in amis:
        prenom, ami = duo  # Séparer les deux prénoms dans chaque duo

        # Ajouter l'ami à la liste de la personne si elle n'y est pas déjà
        if prenom not in dico_reseau:
            dico_reseau[prenom] = [ami]  # Créer une nouvelle entrée si la personne n'est pas encore dans le dictionnaire
        elif ami not in dico_reseau[prenom]:
            dico_reseau[prenom].append(ami)  # Ajouter l'ami à la liste si ce n'est pas un doublon

        # Faire la même chose pour l'ami en inversant les rôles
        if ami not in dico_reseau:
            dico_reseau[ami] = [prenom]  # Créer une nouvelle entrée si l'ami n'est pas encore dans le dictionnaire
        elif prenom not in dico_reseau[ami]:
            dico_reseau[ami].append(prenom)  # Ajouter la personne à la liste de l'ami si ce n'est pas un doublon

    return dico_reseau  # Retourner le dictionnaire d'amitié

# Exemple de données de test
amis = [['Giedrius', 'Mady'], ['Kirsa', 'Vittore'], ['Barbra', 'Faizel'], ['Vittorio', 'Vittore'], ['Louis', 'Placide'], 
        ['Vittorio', 'Barbra'], ['Cloe', 'Björn'], ['Vittore', 'Barbra'], ['Mady', 'Rufino'], ['Cloe', 'Vittorio'], 
        ['Vittorio', 'Placide'], ['Olavi', 'Rufino'], ['Vittore', 'Cloe'], ['Teodor', 'Cloe'], ['Louis', 'Cloe'], 
        ['Idelle', 'Mady'], ['Illtyd', 'Giedrius'], ['Teodor', 'Olavi'], ['Vittore', 'Mady'], ['Rufino', 'Giedrius'], 
        ['Placide', 'Louis'], ['Louis', 'Vittorio'], ['Giedrius', 'Glenys'], ['Björn', 'Olavi'], ['Faizel', 'Placide'], 
        ['Rufino', 'Illtyd'], ['Kirsa', 'Vittorio'], ['Mady', 'Placide'], ['Illtyd', 'Olavi'], ['Cloe', 'Mady'], 
        ['Rufino', 'Vittorio'], ['Placide', 'Rufino'], ['Giedrius', 'Illtyd'], ['Rufino', 'Teodor'], ['Placide', 'Jakob'], 
        ['Barbra', 'Kirsa'], ['Idelle', 'Cloe'], ['Rufino', 'Faizel'], ['Rufino', 'Louis'], ['Björn', 'Cloe'], 
        ['Vittorio', 'Olavi'], ['Vittorio', 'Mady'], ['Glenys', 'Cloe'], ['Björn', 'Idelle'], ['Mady', 'Björn'], 
        ['Barbra', 'Vittorio'], ['Louis', 'Björn'], ['Mady', 'Vittorio'], ['Olavi', 'Barbra'], ['Björn', 'Giedrius'], 
        ['Faizel', 'Barbra'], ['Marwa', 'Idelle'], ['Placide', 'Olavi'], ['Olavi', 'Placide'], ['Cain', 'Vittorio'], 
        ['Jakob', 'Kirsa'], ['Barbra', 'Marwa'], ['Vittorio', 'Louis'], ['Idelle', 'Illtyd'], ['Louis', 'Giedrius'], 
        ['Mady', 'Barbra'], ['Barbra', 'Rufino'], ['Cloe', 'Barbra'], ['Rufino', 'Björn'], ['Björn', 'Faizel'], 
        ['Cloe', 'Vittore'], ['Vittore', 'Björn'], ['Björn', 'Teodor'], ['Louis', 'Rufino'], ['Olavi', 'Teodor'], 
        ['Teodor', 'Mady'], ['Faizel', 'Olavi'], ['Cloe', 'Idelle'], ['Barbra', 'Louis'], ['Barbra', 'Vittore'], 
        ['Giedrius', 'Vittorio'], ['Placide', 'Cain'], ['Vittore', 'Louis'], ['Mady', 'Jakob'], ['Björn', 'Mady'], 
        ['Placide', 'Cloe'], ['Illtyd', 'Rufino'], ['Olavi', 'Louis']]

# Affichage du réseau d'amitié
print(dico_reseau(amis))
````
